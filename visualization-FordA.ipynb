```python
from statsmodels.tsa.stattools import adfuller, kpss

def test_stationarity(name, series):
    s = series.dropna()
    print(f'== {name} (n={len(s)}) ==')
    # ADF: null = unit root (non-stationary)
    statistic, adf_pvalue = adfuller(series, autolag="AIC", result_object=True)
    adf_p = adf_pvalue
    # KPSS: null = stationary
    statistic, kpss_pvalue = kpss(s, nlags='auto',result_object=True)
    kpss_p = kpss_pvalue
    print(f'ADF p-value: {adf_p:.4f} (H0: unit root)')
    print(f'KPSS p-value: {kpss_p:.4f} (H0: stationary)')
    # Simple combined decision rule
    if adf_p < 0.05 and kpss_p > 0.05:
        print('Decision: Series appears STATIONARY — stop differencing.')
    elif adf_p >= 0.05 and kpss_p <= 0.05:
        print('Decision: Series appears NON-STATIONARY — consider differencing (or transform).')
    elif adf_p < 0.05 and kpss_p <= 0.05:
        print('Conflicting: ADF suggests stationary but KPSS suggests non-stationary. Inspect plots; risk of over-differencing.')
    else:
        print('Inconclusive: use ACF/PACF and visual checks; consider seasonal differencing or transform.')
    print()
# Run the test on the original data and the residuals
test_stationarity("Negative class Series", class_negative.iloc[0,1:])
print("\n")
test_stationarity("Positive class Series", class_positive.iloc[0,1:])
# run_adf_test(residuals, "Residuals")
```

#### Class distances

```python
from aeon.distances import (
    euclidean_pairwise_distance,
    dtw_pairwise_distance
)
```

```python
FordA_train1 = FordA_train.iloc[:,1:].to_numpy(dtype=np.float64)
```

```python
train_euclidean = euclidean_pairwise_distance(FordA_train1)
```

```python
train_euclidean
```

```python
train_dtw = dtw_pairwise_distance(FordA_train1, window=0.1, njobs=-1)
```

```python
upper = np.triu(
    np.ones(train_euclidean.shape, dtype=bool),
    k=1
)

same_class = FordA_y_train[:, None] == FordA_y_train[None, :]

within = train_euclidean[upper & same_class]
between = train_euclidean[upper & ~same_class]

print(f"Within-class mean: {within.mean():.4f}")
print(f"Between-class mean: {between.mean():.4f}")
print(f"Separation ratio: {between.mean() / within.mean():.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.kdeplot(within_dtw, label="Within-class", fill=True, alpha=0.3)
sns.kdeplot(between_dtw, label="Between-class", fill=True, alpha=0.3)

plt.xlabel("DTW Distance")
plt.ylabel("Density")
plt.title("FordA: Within vs Between Class DTW Distance")
plt.legend()
plt.show()
```
