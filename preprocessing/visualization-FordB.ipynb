```python
from statsmodels.tsa.stattools import adfuller, kpss

def test_stationarity(name, series):
    s = series.dropna()
    print(f'== {name} (n={len(s)}) ==')
    # ADF: null = unit root (non-stationary)
    statistic, adf_pvalue = adfuller(series, autolag="AIC", result_object=True)
    adf_p = adf_pvalue
    # KPSS: null = stationary
    statistic, kpss_pvalue = kpss(s, nlags='auto',result_object=True)
    kpss_p = kpss_pvalue
    print(f'ADF p-value: {adf_p:.4f} (H0: unit root)')
    print(f'KPSS p-value: {kpss_p:.4f} (H0: stationary)')
    # Simple combined decision rule
    if adf_p < 0.05 and kpss_p > 0.05:
        print('Decision: Series appears STATIONARY — stop differencing.')
    elif adf_p >= 0.05 and kpss_p <= 0.05:
        print('Decision: Series appears NON-STATIONARY — consider differencing (or transform).')
    elif adf_p < 0.05 and kpss_p <= 0.05:
        print('Conflicting: ADF suggests stationary but KPSS suggests non-stationary. Inspect plots; risk of over-differencing.')
    else:
        print('Inconclusive: use ACF/PACF and visual checks; consider seasonal differencing or transform.')
    print()
# Run the test on the original data and the residuals
test_stationarity("Train Negative class Series", train_class_negative.iloc[0,1:])
print("\n")
test_stationarity("Train Positive class Series", train_class_positive.iloc[0,1:])
print("\n")
test_stationarity("Test Negative class Series", test_class_negative.iloc[0,1:])
print("\n")
test_stationarity("Test Positive class Series", test_class_positive.iloc[0,1:])
# run_adf_test(residuals, "Residuals")
```

#### Class distances

```python
from aeon.distances import (
    euclidean_pairwise_distance,
    dtw_pairwise_distance
)
```

```python
FordB_train1 = FordB_train.iloc[:,1:].to_numpy()
FordB_test1 = FordB_test.iloc[:,1:].to_numpy()
```

```python
train_euclidean = euclidean_pairwise_distance(FordB_train1)
```

```python
test_euclidean = euclidean_pairwise_distance(FordB_test1)
```

```python
train_upper = np.triu(
    np.ones(train_euclidean.shape, dtype=bool),
    k=1
)
```

```python
test_upper = np.triu(
    np.ones(test_euclidean.shape, dtype=bool),
    k=1
)
```

```python

same_class = FordB_y_train.to_numpy()[:, None] == FordB_y_train.to_numpy()[None, :]

within = train_euclidean[train_upper & same_class]
between = train_euclidean[train_upper & ~same_class]

print(f"Within-class mean: {within.mean():.4f}")
print(f"Between-class mean: {between.mean():.4f}")
print(f"Separation ratio: {between.mean() / within.mean():.4f}")
```

```python

same_class = FordB_y_test.to_numpy()[:, None] == FordB_y_test.to_numpy()[None, :]

within = test_euclidean[test_upper & same_class]
between = test_euclidean[test_upper & ~same_class]

print(f"Within-class mean: {within.mean():.4f}")
print(f"Between-class mean: {between.mean():.4f}")
print(f"Separation ratio: {between.mean() / within.mean():.4f}")
```

```python
train_dtw = dtw_pairwise_distance(FordB_train1, window=0.01, n_jobs=-1)
```

```python
import seaborn as sns
# Sort training samples by class label
y = FordB_y_train.to_numpy().ravel()
order = np.argsort(y)

# Reorder both rows and columns
D_sorted = train_dtw[np.ix_(order, order)]
mean = np.mean(D_sorted)
std = np.std(D_sorted)

D_normalized = (D_sorted - mean) / std
# Find where class -1 ends and class +1 begins
boundary = np.sum(y == -1)

# Plot
fig, ax = plt.subplots(figsize=(9, 8))

im = ax.imshow(D_normalized, cmap="RdBu_r", aspect="equal")

fig.colorbar(im, ax=ax, label="DTW Distance")

# Draw class boundaries
ax.axvline(boundary - 0.5, color="red", linewidth=1.5)
ax.axhline(boundary - 0.5, color="red", linewidth=1.5)

ax.set_xlabel("Training samples")
ax.set_ylabel("Training samples")
ax.set_title("FordA DTW Distance Matrix (window=0.01)")

plt.show()
```

```python

y = FordB_y_train.to_numpy()

within_dtw = []
between_dtw = []

# Compare each pair of training samples
for i in range(len(y)):
    for j in range(i + 1, len(y)):

        distance = train_dtw[i, j]

        if y[i] == y[j]:
            within_dtw.append(distance)
        else:
            between_dtw.append(distance)

# Calculate mean distances
within_mean = np.mean(within_dtw)
between_mean = np.mean(between_dtw)

print(f"Within-class DTW mean: {within_mean:.4f}")
print(f"Between-class DTW mean: {between_mean:.4f}")
print(f"Separation ratio: {between_mean / within_mean:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.kdeplot(within_dtw, label="Within-class", fill=True, alpha=0.3)
sns.kdeplot(between_dtw, label="Between-class", fill=True, alpha=0.3)

plt.xlabel("DTW Distance")
plt.ylabel("Density")
plt.title("FordA: Within vs Between Class DTW Distance")
plt.legend()
plt.show()
```
